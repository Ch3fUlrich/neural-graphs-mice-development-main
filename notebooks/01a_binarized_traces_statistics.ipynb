## Finding outlier ROIs


```python
def find_outlier_rois(df, column_name):
    """Detect outliers in a dataframe column using the Tukey's fences method."""
    # Calculate the quartiles for the column values
    q1 = df[column_name].quantile(0.25)
    q3 = df[column_name].quantile(0.75)

    # Calculate the interquartile range (IQR)
    iqr = q3 - q1

    # Define the inner and outer (upper) fences
    inner_fence = q3 + 1.5 * iqr
    outer_fence = q3 + 3 * iqr

    # Find the indices of the outliers
    inner_fence_outlier_indices = np.where(df[column_name] > inner_fence)[0]
    outer_fence_outlier_indices = np.where(df[column_name] > outer_fence)[0]

    # Create a sub-dataframe with the outliers
    inner_fence_outliers = df.iloc[inner_fence_outlier_indices]
    outer_fence_outliers = df.iloc[outer_fence_outlier_indices]

    # Keep only the `animal_id` , `date`, and `roi` columns
    inner_fence_outliers = inner_fence_outliers[["animal_id", "date", "roi"]]
    outer_fence_outliers = outer_fence_outliers[["animal_id", "date", "roi"]]

    # Remove duplicates from the dataframes
    inner_fence_outliers = inner_fence_outliers.drop_duplicates()
    outer_fence_outliers = outer_fence_outliers.drop_duplicates()

    # Make sure `roi` is an integer
    inner_fence_outliers["roi"] = inner_fence_outliers["roi"].astype(int)
    outer_fence_outliers["roi"] = outer_fence_outliers["roi"].astype(int)

    return inner_fence_outliers, outer_fence_outliers
```

### Pulse rate outliers


```python
pulse_rate_inner_fence_outliers, pulse_rate_outer_fence_outliers = find_outlier_rois(
    pulse_rates, "pulse_rate"
)
pulse_rate_outer_fence_outliers.head()
```

```python
pct_of_inner_outliers = (len(pulse_rate_inner_fence_outliers) / len(pulse_rates)) * 100
pct_of_outer_outliers = (len(pulse_rate_outer_fence_outliers) / len(pulse_rates)) * 100
print(f"Percentage of inner fence outliers: {pct_of_inner_outliers:.2f}%")
print(f"Percentage of outer fence outliers: {pct_of_outer_outliers:.2f}%")
```
